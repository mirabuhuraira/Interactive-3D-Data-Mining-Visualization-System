```python
import pandas as pd
import plotly.graph_objs as go
import numpy as np
import matplotlib.pyplot as plt
from plotly.offline import plot

df = pd.read_csv('/kaggle/input/datasets/mirabuhuraira/sales-data/sales_data.csv')


def create_interactive_3d_cube_plot(dataframe):
    """
    Create an interactive 3D plot where each data point is represented as a cube.
    """
    # Print column names for debugging
    print("DataFrame Columns:", list(dataframe.columns))

    # Create figure
    fig = go.Figure()

    # Generate a distinct color for each product category
    unique_categories = dataframe['product_category'].unique()
    print(unique_categories)
    cmap = plt.cm.get_cmap('tab20', len(dataframe['product_category'].unique()))  # Adjust number of colors based on unique categories

    # Map each category to a color from the colormap
    color_mapping = {
        category: f'rgba({int(cmap(i)[0] * 255)}, {int(cmap(i)[1] * 255)}, {int(cmap(i)[2] * 255)}, 1)'
        for i, category in enumerate(dataframe['product_category'].unique())
    }
    # Product Category Mapping
    category_mapping = {category : idx for idx, category in enumerate(unique_categories)}
    dataframe['category_numeric'] = dataframe['product_category'].map(category_mapping)

   # Extracting date
    dataframe['date'] = pd.to_datetime(dataframe['date'])

   # Store id Mapping
    unique_stores = dataframe['store_id'].unique()
    store_mapping = {store:idx for idx, store in enumerate(unique_stores)}
    dataframe['store_numeric'] = dataframe['store_id'].map(store_mapping)


    # Prepare hover texts in advance
    hover_texts = []
    for idx, row in dataframe.iterrows():
        # Prepare hover text with all available columns
        hover_text = "Transaction Details:<br>"
        for col in dataframe.columns:
            hover_text += f"{col}: {row[col]}<br>"
        hover_texts.append(hover_text)
        # print(hover_text)
    # print(hover_texts)

    # Add cubes for each data point
    for idx, row in dataframe.iterrows():
        # Use coordinates for plotting
        x = row['date'].year
        y = row['store_numeric']
        z = row['category_numeric']

        # Offset
        offset = 0.3

        # Create cube
        cube = go.Mesh3d(
            # Cube vertices with slight offset to prevent perfect overlap
            x=[x, x, x+offset, x+offset, x, x, x+offset, x+offset],
            y=[y, y+offset, y+offset, y, y, y+offset, y+offset, y],
            z=[z, z, z, z, z+offset, z+offset, z+offset, z+offset],

            # Indices to create cube faces
            i=[7, 0, 0, 0, 4, 4, 6, 6, 4, 0, 3, 2],
            j=[3, 4, 1, 2, 5, 6, 5, 2, 0, 1, 6, 3],
            k=[0, 7, 2, 3, 6, 7, 1, 1, 5, 5, 7, 6],

            color=color_mapping[row['product_category']],
            opacity=0.7,
            # flatshading=True,
            # Use pre-prepared hover text
            text=hover_texts[idx],
            hoverinfo='text'
        )
        fig.add_trace(cube)

        # Customize layout
        fig.update_layout(
            title='3D Cube Visualization of Transaction Data with Categories',
            scene=dict(
                xaxis=dict(
                    title='Year',
                    tickvals=[str(year) for year in dataframe['date'].dt.year.unique()],
                    ticktext=[str(year) for year in dataframe['date'].dt.year.unique()]
                ),
                yaxis=dict(
                    title='Store ID',
                    type='category',  # Set type as 'category' for store IDs
                    tickvals=list(store_mapping.values()),  # Use numeric values for position
                    ticktext=list(store_mapping.keys())  # Display store IDs as text
                ),
                zaxis=dict(
                    title='Product Category',
                    type='category',  # Set type as 'category' for product categories
                    tickvals=list(category_mapping.values()),  # Use numeric values for position
                    ticktext=list(category_mapping.keys())  # Display categories as text
                ),
                aspectmode='manual',
                aspectratio=dict(x=2, y=2, z=1)
            ),
            width=800,
            height=800,
            margin=dict(r=10, b=10, l=10, t=10)
        )

    return fig

# Create and display the plot
fig = create_interactive_3d_cube_plot(df)
fig.show()

# Save as an HTML file
plot(fig, filename='/content/3d_cube_plot.html')
```

```python
import pandas as pd
import plotly.graph_objs as go
import numpy as np
import matplotlib.pyplot as plt

# Load your data
df = pd.read_csv('/kaggle/input/datasets/mirabuhuraira/sales-data/sales_data.csv')

def create_interactive_3d_cube_plot(dataframe):
    """
    Create an interactive 3D plot where each data point is represented as a cube.
    """
    # Print column names for debugging
    print("DataFrame Columns:", list(dataframe.columns))

    # Create figure
    fig = go.Figure()

    # Generate a distinct color for each product category
    unique_categories = dataframe['product_category'].unique()
    cmap = plt.cm.get_cmap('tab10', len(unique_categories))  # Adjust colors based on unique categories

    # Map each category to a color from the colormap
    color_mapping = {
        category: f'rgba({int(cmap(i)[0] * 255)}, {int(cmap(i)[1] * 255)}, {int(cmap(i)[2] * 255)}, 0.7)'
        for i, category in enumerate(unique_categories)
    }

    # Map product categories to numeric values
    category_mapping = {category: idx for idx, category in enumerate(unique_categories)}
    dataframe['category_numeric'] = dataframe['product_category'].map(category_mapping)

    # Convert 'date' column to datetime
    dataframe['date'] = pd.to_datetime(dataframe['date'])

    # Extract year and month from the date
    dataframe['year'] = dataframe['date'].dt.year
    dataframe['month'] = dataframe['date'].dt.month

    # Sort the dataframe by year and month
    dataframe.sort_values(by=['year', 'month'], inplace=True)

    # Prepare hover texts
    hover_texts = []
    for idx, row in dataframe.iterrows():
        hover_text = "Transaction Details:<br>"
        for col in dataframe.columns:
            hover_text += f"{col}: {row[col]}<br>"
        hover_texts.append(hover_text)

    # Add cubes for each data point
    for idx, row in dataframe.iterrows():
        x = row['year']
        y = row['month']  # Use month for the y-axis
        z = row['category_numeric']

        offset = 0.3  # Offset for cube size

        # Create cube
        cube = go.Mesh3d(
            x=[x, x, x + offset, x + offset, x, x, x + offset, x + offset],
            y=[y, y + offset, y + offset, y, y, y + offset, y + offset, y],
            z=[z, z, z, z, z + offset, z + offset, z + offset, z + offset],
            i=[7, 0, 0, 0, 4, 4, 6, 6, 4, 0, 3, 2],
            j=[3, 4, 1, 2, 5, 6, 5, 2, 0, 1, 6, 3],
            k=[0, 7, 2, 3, 6, 7, 1, 1, 5, 5, 7, 6],
            color=color_mapping[row['product_category']],
            opacity=0.7,
            flatshading=True,
            text=hover_texts[idx],
            hoverinfo='text'
        )
        fig.add_trace(cube)

    # Customize layout
    fig.update_layout(
        title='3D Cube Visualization of Transaction Data with Categories',
        scene=dict(
            xaxis=dict(
                title='Year',
                type='category',
                tickvals=list(sorted(dataframe['year'].unique())),  # Sorted years
                ticktext=list(sorted(dataframe['year'].unique()))
            ),
            yaxis=dict(
                title='Month',
                type='category',
                tickvals=list(range(1, 13)),  # Months from 1 to 12
                ticktext=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
            ),
            zaxis=dict(
                title='Product Category',
                type='category',
                tickvals=list(category_mapping.values()),
                ticktext=list(category_mapping.keys())
            ),
            aspectmode='manual',
            aspectratio=dict(x=1, y=2, z=1)
        ),
        width=800,
        height=800,
        margin=dict(r=20, b=20, l=20, t=40)
    )

    return fig

# Create and display the plot
fig = create_interactive_3d_cube_plot(df)
fig.show()
```

```python
import pandas as pd
import plotly.graph_objs as go
import numpy as np
import matplotlib.pyplot as plt

# Load your data
df = pd.read_csv('/kaggle/input/datasets/mirabuhuraira/sales-data/sales_data.csv')

def create_interactive_3d_cube_plot(dataframe):
    """
    Create an interactive 3D plot where each data point is represented as a cube.
    """
    # Print column names for debugging
    print("DataFrame Columns:", list(dataframe.columns))

    # Create figure
    fig = go.Figure()

    # Generate a distinct color for each product category
    unique_categories = dataframe['product_category'].unique()
    cmap = plt.cm.get_cmap('tab10', len(unique_categories))  # Adjust colors based on unique categories

    # Map each category to a color from the colormap
    color_mapping = {
        category: f'rgba({int(cmap(i)[0] * 255)}, {int(cmap(i)[1] * 255)}, {int(cmap(i)[2] * 255)}, 0.7)'
        for i, category in enumerate(unique_categories)
    }

    # Map product categories and store IDs to numeric values
    category_mapping = {category: idx for idx, category in enumerate(unique_categories)}
    dataframe['category_numeric'] = dataframe['product_category'].map(category_mapping)

    unique_stores = dataframe['store_id'].unique()
    store_mapping = {store: idx for idx, store in enumerate(unique_stores)}
    dataframe['store_numeric'] = dataframe['store_id'].map(store_mapping)

    # Convert 'date' column to datetime
    dataframe['date'] = pd.to_datetime(dataframe['date'])

    # Extract quarter (Q1, Q2, Q3, Q4) from the 'date' column
    dataframe['quarter'] = dataframe['date'].dt.quarter  # Get the quarter as an integer (1, 2, 3, 4)

    # Prepare hover texts
    hover_texts = []
    for idx, row in dataframe.iterrows():
        hover_text = "Transaction Details:<br>"
        for col in dataframe.columns:
            hover_text += f"{col}: {row[col]}<br>"
        hover_texts.append(hover_text)

    # Add cubes for each data point
    for idx, row in dataframe.iterrows():
        x = row['date'].year
        y = row['quarter']  # Use the quarter (as integer)
        z = row['category_numeric']

        offset = 0.3  # Offset for cube size

        # Create cube
        cube = go.Mesh3d(
            x=[x, x, x + offset, x + offset, x, x, x + offset, x + offset],
            y=[y, y + offset, y + offset, y, y, y + offset, y + offset, y],
            z=[z, z, z, z, z + offset, z + offset, z + offset, z + offset],
            i=[7, 0, 0, 0, 4, 4, 6, 6, 4, 0, 3, 2],
            j=[3, 4, 1, 2, 5, 6, 5, 2, 0, 1, 6, 3],
            k=[0, 7, 2, 3, 6, 7, 1, 1, 5, 5, 7, 6],
            color=color_mapping[row['product_category']],
            opacity=0.7,
            flatshading=True,
            text=hover_texts[idx],
            hoverinfo='text'
        )
        fig.add_trace(cube)

    # Customize layout
    fig.update_layout(
        title='3D Cube Visualization of Transaction Data with Categories',
        scene=dict(
            xaxis=dict(
                title='Year',
                type='category',
                tickvals=list(dataframe['date'].dt.year.unique()),
                ticktext=list(dataframe['date'].dt.year.unique())
            ),
            yaxis=dict(
                title='Quarter',
                type='category',
                tickvals=[1, 2, 3, 4],  # Quarters from 1 to 4
                ticktext=['Q1', 'Q2', 'Q3', 'Q4']  # Quarter labels
            ),
            zaxis=dict(
                title='Product Category',
                type='category',
                tickvals=list(category_mapping.values()),
                ticktext=list(category_mapping.keys())
            ),
            aspectmode='manual',
            aspectratio=dict(x=1, y=2, z=1)
        ),
        width=800,
        height=800,
        margin=dict(r=20, b=20, l=20, t=40)
    )

    return fig

# Create and display the plot
fig = create_interactive_3d_cube_plot(df)
fig.show()
```
